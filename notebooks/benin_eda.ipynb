## Summary statistics and missing value report


```python
import pandas as pd
```

```python
df = pd.read_csv("../data/benin/raw/benin-malanville.csv")

#thd rows and columns of the dataframe
print(f"Shape (rows, columns): {df.shape}\n")

#the data type of each column
print("Column dtypes:")
print(df.dtypes, "\n")

# Null counts and columns with >5% nulls
na_counts = df.isna().sum()
print("Null counts per column:")
print(na_counts)

percent_null = (na_counts / len(df)) * 100
print("\nNull percentage per column:")
print(percent_null.round(2))

cols_over_5pct = percent_null[percent_null > 5].sort_values(ascending=False)
if not cols_over_5pct.empty:
    print("\nColumns with >5% nulls:")
    for col, pct in cols_over_5pct.items():
        print(f"- {col}: {pct:.2f}% ({na_counts[col]} nulls)")
else:
    print("\nNo columns with >5% nulls.")

#descsriptive statistics
print("\nDescriptive statistics (numeric):")
display(df.describe())
print("\nDescriptive statistics (all columns):")
display(df.describe(include='all'))
```

# Data cleaning

## Data type conversion: fix the date column to proper datetime format

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'], errors='coerce')
#check the column data types 
print("Column dtypes:")
print(df.dtypes, "\n")
```

## Catching any missing values- empty strings, 'NA', -999 sentinels

```python
# replace any blank or empty strings with NaN
df.replace({r'^\s*$': pd.NA, r'(?i)^NA$': pd.NA}, regex=True, inplace=True)
# replace numeric sentinels
df.replace([-999, -999.0], pd.NA, inplace=True)
print("\nAfter replacing sentinels and blanks with NaN:")
na_counts_after = df.isna().sum()
print(na_counts_after)
```

## Remove any incorrect entries based on domain knowledge 

```python
#check for negative values in GHI, DNI, DHI, ModA, ModB, WS, WSgust since irradiance should be non-negative and wind speed cannot be negative
cols = ['GHI', 'DNI', 'DHI',"ModA","ModB","WS","WSgust"]
neg_counts = (df[cols] < 0).sum()
neg_pct = (neg_counts / len(df) * 100).round(2)
print("Negative counts and percentage (of total rows):")
for c in cols:
    print(f"- {c}: {neg_counts[c]} ({neg_pct[c]}%)")
print("total negatives:", neg_counts.sum())

#find the hours when negatives occur
for col in cols:
    mask = df[col] < 0
    if mask.any():
        neg_hours = df.loc[mask, 'Timestamp'].dt.hour.value_counts().sort_index()
        print(f"- negatives by hour count for {col} negatives: timestamp={neg_hours}")
    else:
        print(f"- no negatives for {col}")

```

```python
# replace all negative values in GHI, DNI, DHI with 0
df['GHI'] = df['GHI'].clip(lower=0)
df['DNI'] = df['DNI'].clip(lower=0) 
df['DHI'] = df['DHI'].clip(lower=0)
print("total negatives after clipping to 0:", (df[cols] < 0).sum().sum())
df.describe()
```

```python
#further data validation checks
print("Check for incorrect values")
def check_sensor_data(df: pd.DataFrame):
    required_cols = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'WS', 'WSgust']
    missing = [col for col in required_cols if col not in df.columns]
    
    if missing:
        raise ValueError(f"Missing columns: {missing}")

    issues = {}
   
    # GHI, DNI, DHI checks
    #since negatives have now been clipped to 0, we just check for any remaining negatives
    for col in ['GHI', 'DNI', 'DHI']:
        issues[f'{col}_negative'] = df[df[col] < 0].index.tolist()
    
    # DHI should not exceed GHI 
    issues['DHI_gt_GHI'] = df[df['DHI'] > df['GHI']].index.tolist()

    # GHI, DNI, DHI should not exceed ~1300 W/m²
    # This is based on the maximum expected solar irradiance at sea level under clear sky conditions
    issues['GHI_above_1300'] = df[df['GHI'] > 1300].index.tolist()
    issues['DNI_above_1300'] = df[df['DNI'] > 1300].index.tolist()
    issues['DHI_above_1300'] = df[df['DHI'] > 1300].index.tolist()

    # Sensor irradiance checks (ModA, ModB)
    # These should be non-negative and not exceed ~1300 W/m²
    for col in ['ModA', 'ModB']:
        issues[f'{col}_negative'] = df[df[col] < 0].index.tolist()
        issues[f'{col}_above_1300'] = df[df[col] > 1300].index.tolist()

    # Wind speed (WS, WSgust) checks
    for col in ['WS', 'WSgust']:
        issues[f'{col}_negative'] = df[df[col] < 0].index.tolist()
        issues[f'{col}_above_40'] = df[df[col] > 40].index.tolist()  # adjust for your climate

    # WSgust should not be less than WS
    issues['WSgust_lt_WS'] = df[df['WSgust'] < df['WS']].index.tolist()

    # Summary report
    for key, idxs in issues.items():
        if idxs:
            print(f"{key}: {len(idxs)} at rows {idxs[:5]}{'...' if len(idxs) > 5 else ''}")

    return issues
issues = check_sensor_data(df)
```

## Check for outliers  in GHI,DNI,DHI,ModA,ModB,WS, WSgust


```python
#checking for outliers using Z-score method
print("Calculating outliers")
from scipy.stats import zscore
import numpy as np

def flag_zscore_outliers(df: pd.DataFrame, columns: list) -> pd.DataFrame:
    z_scores = df[columns].apply(zscore)

    # Create a boolean mask of rows where any Z-score > 3 in magnitude
    outlier_mask = (np.abs(z_scores) > 3)
    df['z_outlier'] = outlier_mask.any(axis=1)

    print(f"Flagged {df['z_outlier'].sum()} outlier rows (|Z| > 3).")
# Define the relevant columns
cols_to_check = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'WS', 'WSgust']

# Step 2: Flag Z-score outliers
df_with_outliers = flag_zscore_outliers(df, cols_to_check)
df_with_outliers
```

```python
#outlier cleaning loop
from scipy.stats import zscore
import numpy as np
print("Initial Outlier Check Befor Cleaning")
print("Calculating outliers")
cols_to_check = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'WS', 'WSgust']

def zscore_outliers_mask(df: pd.DataFrame, columns: list) -> pd.DataFrame:
    z_scores = df[columns].apply(zscore)
    # Create a boolean mask of rows where any Z-score > 3 in magnitude
    outlier_mask = (np.abs(z_scores) > 3)
    return outlier_mask

def get_no_of_outliers(df: pd.DataFrame, columns: list) -> pd.DataFrame:
    z_scores = df[columns].apply(zscore)

    # Create a boolean mask of rows where any Z-score > 3 in magnitude
    outlier_mask = (np.abs(z_scores) > 3)
    df['z_outlier'] = outlier_mask.any(axis=1)
    return df['z_outlier'].sum()

no_of_outliers = get_no_of_outliers(df, cols_to_check)

while no_of_outliers > 0:
    print("Cleaning outliers with imputation")

    outlier_mask = zscore_outliers_mask(df, cols_to_check)
    for col in cols_to_check:
        df.loc[outlier_mask[col], col] = df[col].median()
    
    no_of_outliers = get_no_of_outliers(df, cols_to_check)
    print("number of outliers: ", no_of_outliers)

print("Final Outlier Check After Cleaning")
df_with_outliers = flag_zscore_outliers(df, cols_to_check)
df_with_outliers
```

```python
#check for incorrect values after outlier cleaning
print("Check for incorrect values after outlier cleaning")
issues = check_sensor_data(df)

# Fix cases where DHI > GHI by setting DHI = GHI
dhi_gt_mask = df['DHI'] > df['GHI']
if dhi_gt_mask.any():
    print(f"Found {dhi_gt_mask.sum()} rows with DHI > GHI. Setting DHI = GHI for those rows.")
    df.loc[dhi_gt_mask, 'DHI'] = df.loc[dhi_gt_mask, 'GHI']
else:
    print('No rows with DHI > GHI found.')

# Fix cases where WSgust < WS by setting WSgust = WS
wsg_lt_mask = df['WSgust'] < df['WS']
if wsg_lt_mask.any():
    print(f"Found {wsg_lt_mask.sum()} rows with WSgust < WS. Setting WSgust = WS for those rows.")
    df.loc[wsg_lt_mask, 'WSgust'] = df.loc[wsg_lt_mask, 'WS']
else:
    print('No rows with WSgust < WS found.')

# Final check for incorrect values
print("Final check for incorrect values after fixes")
issues = check_sensor_data(df)
issues

```

```python
#dropping the comments column and the z_outlier column
df = df.drop(columns=['Comments', 'z_outlier'])
df.describe(include='all')
```

## Time Series Analysis

```python
cleaned_df = pd.read_csv("../data/benin/processed/benin-cleaned.csv")
len(cleaned_df)
```

```python
# Fast multiseries time plot (GHI, DNI, DHI, Tamb) with automatic downsampling/resampling
# Ensures Timestamp is datetime/indexed, automatically chooses a resample interval to keep the
# plotted points small and fast to render.
# function always computes a reasonable time-bucket when needed.
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np


def plot_timeseries_fast(df: pd.DataFrame, cols: list | None = None, max_points: int = 5000):
    """Plot multiple time series vs Timestamp with automatic downsampling.

    The function automatically computes a time-based resample rule (or uniform sampling fallback)
    so you do not need to pass resample rules. It prints the rule used and plots the series.
    """
    df = df.copy()
    if 'Timestamp' not in df.columns:
        raise ValueError('DataFrame must contain a Timestamp column')

    # ensure datetime dtype
    if not pd.api.types.is_datetime64_any_dtype(df['Timestamp']):
        df['Timestamp'] = pd.to_datetime(df['Timestamp'], errors='coerce')

    # drop rows without valid timestamp and sort
    df = df.dropna(subset=['Timestamp']).sort_values('Timestamp')
    df = df.set_index('Timestamp')

    # default columns if not provided
    if cols is None:
        pref = ['GHI', 'DNI', 'DHI', 'Tamb']
        cols = [c for c in pref if c in df.columns]
    else:
        cols = [c for c in cols if c in df.columns]

    if not cols:
        raise ValueError('No requested columns found in dataframe to plot')

    n = len(df)

    # Choose a time-based resample rule to reduce points to roughly max_points
    span_seconds = (df.index.max() - df.index.min()).total_seconds() if n > 0 else 0
    span_hours = span_seconds / 3600.0 if span_seconds is not None else 0

    if n == 0:
        raise ValueError('DataFrame has no rows with valid timestamps')

    if span_hours > 0:
        # compute hours per bin required to get ~max_points points
        hours_per_bin = max(1, int(round(span_hours / max_points)))
        resample_rule = f"{hours_per_bin}h"
        df_small = df[cols].resample(resample_rule).mean().dropna(how='all')
        used_rule = resample_rule
    else:
        # fallback uniform sampling
        step = max(1, n // max_points)
        df_small = df[cols].iloc[::step]
        used_rule = f"sample_every={step}"

    print(f"Original rows={n}, plotted points per series~={len(df_small)}; resample_rule={used_rule}")

    # plot each series as a subplot
    axes = df_small.plot(subplots=True, figsize=(12, 3 * len(df_small.columns)), grid=True, linewidth=0.8)


    plt.tight_layout()
    plt.show()


# Ensure cleaned_df is available (read file if not in memory) and call the helper
try:
    _ = cleaned_df
except NameError:
    cleaned_df = pd.read_csv('../data/benin/processed/benin-cleaned.csv')

plot_timeseries_fast(cleaned_df)
```

## Monthly profile of GHI,DHI,DNI,Tamb

```python
import seaborn as sns
import matplotlib.pyplot as plt

# ensure Timestamp is datetime and Month exists
cleaned_df['Timestamp'] = pd.to_datetime(cleaned_df['Timestamp'])
if 'Month' not in cleaned_df.columns:
    cleaned_df['Month'] = cleaned_df['Timestamp'].dt.month

cols = ['GHI', 'DNI', 'DHI', 'Tamb']
for col in cols:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Month', y=col, data=cleaned_df)
    plt.title(f'Monthly Distribution of {col}')
    plt.xlabel('Month')
    # set a sensible ylabel
    if col in ['GHI', 'DNI', 'DHI']:
        plt.ylabel(f'{col} (W/m²)')
    else:
        plt.ylabel(col)
    plt.grid(True)
    plt.show()

```

## Daily and hourly profile of GHI,DHI,DNI,Tamb

```python
cleaned_df['Day'] = cleaned_df['Timestamp'].dt.day
df['Hour'] = df['Timestamp'].dt.hour

variables = ['GHI', 'DNI', 'DHI', 'Tamb']

for var in variables:
    print(f"Plotting {var} by day")
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Day', y=var, data=cleaned_df)
    plt.title(f'Daily Distribution of {var}')
    plt.xlabel('Day')
    if( var in ['GHI', 'DNI', 'DHI']):
        plt.ylabel(f'{var} (W/m²)')
    else:
        plt.ylabel(f'{var}(°C)')
    plt.grid(True)
    plt.show()
    
    print("------------------------------------------------")
    print(f"Plotting average hourly pattern for {var}")
    hourly_mean = df.groupby('Hour')[var].mean()
    plt.plot(hourly_mean)
    plt.xlabel('Hour of Day')
    if( var in ['GHI', 'DNI', 'DHI']):
        plt.ylabel(f'Average{var} (W/m²)')
    else:
        plt.ylabel(f'Average{var}(°C)')
    
    plt.title(f'Average {var} by Hour')
    plt.show()
```
